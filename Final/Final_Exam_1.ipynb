```python
# Import data

# (none)


# Data Representation

# A single expense will be stored as a dictionary:
# category (string)
# vendor (string)
# amount (float)
# date (string)

# All expenses will be stored in a list called expenses


# Function Definitions

# 1) define add_expense(expenses):

# prompt user for category
# prompt user for vendor
# prompt user for amount
# prompt user for date
# store values in a dictionary
# append dictionary to expenses list

# 2) define filter_by_category(expenses, category):

# loop through the expenses
# if expense category matches input category
# add amount to running total
# return total

# 3) define filter_by_vendor(expenses, vendor):

# loop through expenses
# if expense vendor matches input vendor
# add amount to running total
# return total

# 4) define total_expenses(expenses):

# loop through all expenses
# sum all amounts
# return total


# Program Logic

# initialize empty list called expenses

# while True:
# display main menu:
# 1 Add New Expense
# 2 Get Totals
# 3 Exit

# if user selects 1:
# call add_expense(expenses)
# if user selects 2:
# show sub menu:
# A: All Expenses
# C: Filter by Category
# V: Filter by Vendor
# if user selects 3:
# break loop and exit program

def format_money(value):
    return f"${value:.2f}"


def safe_float_input(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid input. Please enter a number.")


def add_expense(expenses):
    category = input("Enter Category: ")
    vendor = input("Enter Vendor: ")

    amount = safe_float_input("Enter Amount: ")
    date = input("Enter Date: ")

    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    expenses.append(expense)


def total_expenses(expenses):
    return sum(expense["amount"] for expense in expenses)


def filter_by_category(expenses, category):
    return sum(expense["amount"] for expense in expenses if expense["category"] == category)


def filter_by_vendor(expenses, vendor):
    return sum(expense["amount"] for expense in expenses if expense["vendor"] == vendor)


while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == "1":
        add_expense(expenses)

    elif choice == "2":
        print("--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter by Vendor")

        sub_choice = input("Choice: ")

        if sub_choice == "A":
            print("Total:", format_money(total_expenses(expenses)))

        elif sub_choice == "C":
            cat = input("Enter Category to sum: ")
            print("Total:", format_money(filter_by_category(expenses, cat)))

        elif sub_choice == "V":
            ven = input("Enter Vendor to sum: ")
            print("Total:", format_money(filter_by_vendor(expenses, ven)))

    elif choice == "3":
        print("Goodbye!")
        break
```
